```python
def get_image_paths(folder):
    """Get all image file paths from a folder."""
    if not os.path.exists(folder):
        return []
    return [
        os.path.join(folder, f)
        for f in os.listdir(folder)
        if f.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp', '.tiff', 'webp'))
    ]
```
